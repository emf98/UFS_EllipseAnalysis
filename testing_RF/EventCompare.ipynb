### Simple date comparison. 

File generated 12/12/2025

```python
#imports
import numpy as np 
import collections
import pickle
```

```python
##comparison definition
def all_compare_keys(list1, list2, list3, list4, list5):
    #count for list 1
    count1 = collections.Counter() 
    for day in list1:
        count1[day] += 1
    #count for list 2
    count2 = collections.Counter() 
    for day in list2:
        count2[day] += 1
    #count for list 3
    count3 = collections.Counter() 
    for day in list3:
        count3[day] += 1
    #count for list 4
    count4 = collections.Counter() 
    for day in list4:
        count4[day] += 1
    #count for list 5
    count5 = collections.Counter() 
    for day in list5:
        count5[day] += 1
        
    result = []
    count1_only = []
    #begin comparing dictionaries. 
    for key in count1:
        if key in count2 and key in count3 and key in count4 and key in count5:
            #save the key to look at for composites
            result.append(key)
        if key not in count2 and key not in count3 and key not in count4 and key not in count5:
            count1_only.append(key)
    
    count2_only =[]
    for key in count2:
        if key not in count1 and key not in count3 and key not in count4 and key not in count5:
            count2_only.append(key)
    
    count3_only =[]
    for key in count3:
        if key not in count1 and key not in count2 and key not in count4 and key not in count5:
            count3_only.append(key)

    count4_only =[]
    for key in count4:
        if key not in count1 and key not in count2 and key not in count3 and key not in count5:
            count4_only.append(key)

    count5_only =[]
    for key in count5:
        if key not in count1 and key not in count2 and key not in count3 and key not in count4:
            count5_only.append(key)
    
    return result, count1_only, count2_only, count3_only, count4_only, count5_only;

#______________________________________________________________#
##open files and use comparison code
def import_and_compare(category, region):
    #opening, for example, only false positive files for one region, SEUS.
    infile = open(f'./index/{category}_{region}_ERA5.p',"rb",)
    era5 = pickle.load(infile)
    infile.close()

    infile = open(f'./index/{category}_{region}_Prototype5.p',"rb",)
    P5 = pickle.load(infile)
    infile.close()

    infile = open(f'./index/{category}_{region}_Prototype6.p',"rb",)
    P6 = pickle.load(infile)
    infile.close()

    infile = open(f'./index/{category}_{region}_Prototype7.p',"rb",)
    P7 = pickle.load(infile)
    infile.close()

    infile = open(f'./index/{category}_{region}_Prototype8.p',"rb",)
    P8 = pickle.load(infile)
    infile.close()

    return all_compare_keys(era5, P5, P6, P7, P8)   
```

```python
category = ["pos","positive"]
region = "Europe"
result, count1_only, count2_only, count3_only, count4_only, count5_only = import_and_compare(category[0],region)
print(f'There are {str(len(result))} events shared across all prototypes for {category[1]} events in {region}.')
print(f'{str(len(count1_only))} are only in ERA5.')
print(f'{str(len(count2_only))} are only in Prototype5.')
print(f'{str(len(count3_only))} are only in Prototype6.')
print(f'{str(len(count4_only))} are only in Prototype7.')
print(f'{str(len(count5_only))} are only in Prototype8.')
print('_________________________________________________')

category = ["neg","negative"]
region = "Europe"
result, count1_only, count2_only, count3_only, count4_only, count5_only = import_and_compare(category[0],region)
print(f'There are {str(len(result))} events shared across all prototypes for {category[1]} events in {region}.')
print(f'{str(len(count1_only))} are only in ERA5.')
print(f'{str(len(count2_only))} are only in Prototype5.')
print(f'{str(len(count3_only))} are only in Prototype6.')
print(f'{str(len(count4_only))} are only in Prototype7.')
print(f'{str(len(count5_only))} are only in Prototype8.')
print('_________________________________________________')

category = ["Fpos","false positive"]
region = "Europe"
result, count1_only, count2_only, count3_only, count4_only, count5_only = import_and_compare(category[0],region)
print(f'There are {str(len(result))} events shared across all prototypes for {category[1]} events in {region}.')
print(f'{str(len(count1_only))} are only in ERA5.')
print(f'{str(len(count2_only))} are only in Prototype5.')
print(f'{str(len(count3_only))} are only in Prototype6.')
print(f'{str(len(count4_only))} are only in Prototype7.')
print(f'{str(len(count5_only))} are only in Prototype8.')
print('_________________________________________________')

category = ["Fneg","false negative"]
region = "Europe"
result, count1_only, count2_only, count3_only, count4_only, count5_only = import_and_compare(category[0],region)
print(f'There are {str(len(result))} events shared across all prototypes for {category[1]} events in {region}.')
print(f'{str(len(count1_only))} are only in ERA5.')
print(f'{str(len(count2_only))} are only in Prototype5.')
print(f'{str(len(count3_only))} are only in Prototype6.')
print(f'{str(len(count4_only))} are only in Prototype7.')
print(f'{str(len(count5_only))} are only in Prototype8.')
```
